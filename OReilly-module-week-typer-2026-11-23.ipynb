# Default values

We can define our function that is (indirectly) invoked by Typer with a default argument value.

If you define a default argument value for a Typer-invoked function, then you cannot invoke the program (from the command line) with a positional argument for that parameter. Rather, you must use a keyword argument -- meaning, you say `--name` and then the value.
